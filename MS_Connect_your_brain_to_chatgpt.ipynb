```python
import openai

openai.api_key = "sk-or-xxxxxxxxxxxxxxxxxxxxxxxxxxxxxxxxx"
openai.api_base = "https://openrouter.ai/api/v1"

def generate_problem(prompt):
    try:
        response = openai.ChatCompletion.create(
            model="gpt-3.5-turbo",
            messages=[{"role": "user", "content": prompt}],
            max_tokens=300
        )
        content = response['choices'][0]['message']['content']
        print("Raw Response:\n", content)
        return content
    except Exception as e:
        print(f"Error occurred: {e}")
        return None

def generate_first_problem(topic):
    prompt = f"""
Generate a math problem on the topic: {topic}.
Respond using this format exactly:
[Feedback]: First question. No feedback needed yet.
[Next Problem]: (insert the problem here)
"""
    return generate_problem(prompt)

def generate_next_problem(user_answer, focus_level, prev_problem, topic):
    prompt = f"""
The student answered the last problem as: '{user_answer}'.
Their focus level is {focus_level}/5.
The last problem was: '{prev_problem}'.

Give your response in this exact format:
[Feedback]: (feedback on user's answer)
[Next Problem]: (next math problem, adjust difficulty based on focus level)

Rules:
- If focus is low (1-2), make it easier.
- If focus is medium (3), keep same level.
- If focus is high (4-5), make it slightly harder.
- Keep the language clear and concise.
"""
    return generate_problem(prompt)

def extract_feedback_and_problem(response_text):
    feedback = ""
    next_problem = ""
    try:
        for line in response_text.splitlines():
            if line.startswith("[Feedback]:"):
                feedback = line.replace("[Feedback]:", "").strip()
            elif line.startswith("[Next Problem]:"):
                next_problem = line.replace("[Next Problem]:", "").strip()
    except Exception as e:
        print(f"Error in parsing response: {e}")
    return feedback, next_problem

# Main loop
topic = input("Enter the math topic you want to practice: ")
raw_response = generate_first_problem(topic)
feedback, problem = extract_feedback_and_problem(raw_response)
print("\n", feedback)
print("First Problem:\n", problem)

while True:
    user_answer = input("\nYour Answer: ")
    focus_level = input("Enter focus level (1-5): ")
    raw_response = generate_next_problem(user_answer, focus_level, problem, topic)
    feedback, problem = extract_feedback_and_problem(raw_response)
    print("\nFeedback:", feedback)
    print("Next Problem:\n", problem)
```
